# RunPod Pod for GPCR-Diffleop

This Pod uses a minimal Miniforge image and creates the project environment from `Diffleop/environment.yaml`.

- The image is `condaforge/miniforge3:26.3.2-3`, so `conda` and `mamba` are available immediately after the Pod starts.
- A PyTorch prebuilt RunPod template is intentionally not used. Any PyTorch dependency should come from `Diffleop/environment.yaml`, keeping the runtime closer to the project definition.
- The GPU is `NVIDIA A40`. It provides 48GB VRAM and is a safer fit for the older CUDA stack specified by the project environment than RTX 4090.
- `container-disk-in-gb` is set to 80GB because the conda environment, RDKit, OpenBabel, PyG, and related packages can consume a lot of disk space. 50GB may be tight.
- A 50GB volume is mounted at `/workspace`. Put code, checkpoints, generated outputs, and other persistent files there.
- Port `8888/http` is for Jupyter, and `22/tcp` is for SSH.
- `--docker-args "sleep infinity"` keeps the minimal container running so RunPod can attach a terminal or SSH session.

Expected setup after creating the Pod:

```bash
cd /workspace
git clone <repository-url>
cd gpcr-diffleop/Diffleop
mamba env create -f environment.yaml
conda activate diffleop
```

If `mamba env create` fails to solve the environment, try the same YAML with `conda env create -f environment.yaml`.


````bash
runpodctl pod create \
  --name gpcr-diffleop \
  --image "condaforge/miniforge3:26.3.2-3" \
  --gpu-id "NVIDIA A40" \
  --gpu-count 1 \
  --container-disk-in-gb 80 \
  --volume-in-gb 50 \
  --volume-mount-path /workspace \
  --ports "8888/http,22/tcp" \
  --docker-args "sleep infinity" \
  --cloud-type SECURE
````
